2. Попробуем обучить сеть, изменяя вес в разные стороны и рассчитывая значение ошибки для каждого изменения

```python
weight = 0.5
input = 0.5
goal_prediction = 0.8

step_amount = 10

for iteration in range(1101):

  prediction = input * weight
  error = (prediction - goal_prediction) ** 2

  print("Error:" + str(error) + " Prediction:" + str(prediction))

  up_prediction = input * (weight + step_amount) #попробовать увеличить
  up_error = (goal_prediction - up_prediction) ** 2

  down_prediction = input * (weight - step_amount) #попробовать уменьшить
  down_error = (goal_prediction - down_prediction) ** 2

  if(down_error < up_error):
    weight = weight - step_amount #если уменьшение дало лучший результат, уменьшить

  if(down_error > up_error):
    weight = weight + step_amount # если увеличение дало лучший результат, увеличить
```

3. Реализуйте градиентный спуск: определите направление и величину изменения веса, используя производную функции потерь для текущего веса . Выполните 20 итераций градиентного спуска .
 weight = weight - derivative

```python
# 2.11 Задание
weight = 0.5
goal_pred = 0.8
input = 0.5

for iteration in range(20):
  pred = input * weight
  error = (pred - goal_pred) ** 2

  # вычисляем направление и величину изменения
  # производная mse по весу: 2 * input * (pred - goal_pred)
  direction_and_amount = 2 * input * (pred - goal_pred)

  # обновляем вес
  weight = weight - direction_and_amount

  print(f"Итерация {iteration}: Вес = {weight:.6f}, Прогноз = {pred:.6f}, Ошибка = {error:.6f}")
```

4. Реализуйте алгоритм градиентного спуска, добавив параметр скорости обучения.

```python
weight = 0.5
goal_prediction = 0.8
input = 0.5

alpha = 0.1 # скорость обучения, можно попробовать 10, 1, 0.1, 0.01

for iteration in range(20):
  pred = input * weight
  error = (pred - goal_prediction) ** 2

  # производная
  derivative = 2 * input * (pred - goal_prediction)

  # обновление веса с учетом скорости обучения
  weight = weight - (alpha * derivative)

  print(f"Итерация {iteration}: вес = {weight:.6f}, прогноз = {pred:.6f}, ошибка = {error:.6f}")
```

Что происходит при input=2?

```python
# input = 2 без использования alpha (скорости обучения)
weight = 0.5
goal_pred = 0.8
input_val = 2

print("Вход = 2, Без Alpha")
for iteration in range(5):
    pred = input_val * weight
    error = (pred - goal_pred) ** 2
    derivative = 2 * input_val * (pred - goal_pred)
    weight = weight - derivative
    print(f"Итерация {iteration}: прогноз = {pred:.2f}, ошибка = {error:.2f}")

# input = 2 с использованием очень большой alpha = 10
weight = 0.5
alpha = 10
print("\nВход = 2, Alpha = 10")
for iteration in range(3):
    pred = input_val * weight
    error = (pred - goal_pred) ** 2
    derivative = 2 * input_val * (pred - goal_pred)
    weight = weight - (alpha * derivative)
    print(f"Итерация {iteration}: прогноз = {pred:.2f}, ошибка = {error:.2f}")

# input = 2 с оптимальной alpha = 0.1 или 0.01
weight = 0.5
alpha = 0.1
print("\nВход = 2, Alpha = 0.1")
for iteration in range(10):
    pred = input_val * weight
    error = (pred - goal_pred) ** 2
    derivative = 2 * input_val * (pred - goal_pred)
    weight = weight - (alpha * derivative)
    if iteration % 2 == 0: # выводим каждую вторую итерацию
        print(f"Итерация {iteration}: прогноз = {pred:.4f}, ошибка = {error:.4f}")
```

***input = 2 без alpha***  
Каждое новое изменение веса будет делать прогноз всё дальше и дальше от цели, ошибка будет расти до бесконечности. Это всё потому, что большой вход слишком сильно увеличивает значение производной  

***input = 2 и alpha = 10***  
Ошибка уходит в бесконечность  

***input = 2 и alpha = 0.1 или alpha = 0.01***  
Алгоритм начинаетс сходиться. Скорость обучения подавляет слишком резкие прыжки веса, и сеть постепенно находит правильный ответ, даже если входные данные велики
